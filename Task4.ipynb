```python
import numpy as np

likes = np.array([120, 150, 200, 180, 250, 300, 220])
comments = np.array([10, 15, 20, 12, 25, 30, 18])

engagement = likes + comments
average_engagement = np.mean(engagement)

print("Average Engagement:", average_engagement)
```

```python
import numpy as np

prices = np.array([250, 180, 300, 150, 220])
discounts = np.array([50, 30, 60, 20, 40])

final_price = prices - discounts

print("Final Prices:", final_price)
```

```python
import numpy as np

scores = np.array([175, 190, 165, 210, 180])

result = scores > 180

print("Scores greater than 180:", result)
```

```python
import numpy as np

paytm = np.array([1, 0, 1, 0, 0, 1])
phonepe = np.array([0, 1, 1, 0, 1, 0])

result = np.logical_or(paytm, phonepe)

print("Paid by either app:", result)
```

```python
import numpy as np

steps = np.array([4000, 5000, 6000, 4500, 7000, 5500, 6500])

bonus_steps = steps + 500

total_steps = np.sum(bonus_steps)

print("Steps after bonus:", bonus_steps)
print("Total steps for the week:", total_steps)
```
