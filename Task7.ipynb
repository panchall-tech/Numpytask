```python
import numpy as np

zomato = np.array([4.5, 4.2, 3.8, 4.7, 4.0])
swiggy = np.array([4.1, 4.6, 3.9, 4.3, 4.8])

result = np.concatenate((zomato, swiggy))

print(result)
```

```python
import numpy as np

messages = np.array([
    [20, 25, 30, 22, 28, 35, 40],
    [10, 15, 12, 18, 20, 25, 30],
    [30, 35, 40, 38, 45, 50, 55]
])

new_user = np.array([15, 20, 18, 22, 25, 28, 30])

messages = np.insert(messages, 3, new_user, axis=0)

totals = np.sum(messages, axis=1)

least_user = np.argmin(totals)

messages = np.delete(messages, least_user, axis=0)

print(messages)
```

```python
import numpy as np

views = np.array([100, 200, 300, 400, 500])

view_array = views.view()
copy_array = views.copy()

view_array[0] = 999
copy_array[0] = 888

print("Original:", views)
print("View:", view_array)
print("Copy:", copy_array)
```
