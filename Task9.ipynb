```python
import numpy as np

prices = np.array([299, 499, 799, 0, 1599, -1, 899])

print(prices)
print(prices.dtype)
```

```python
positive_prices = prices[prices > 0]

average_price = positive_prices.mean()

prices[prices <= 0] = average_price

print(prices)
```

```python
quantities = np.array([2, 1, 3, 4, 2, 1, 5])

total_bill = prices * quantities

print(total_bill)
```

```python
print("Minimum:", prices.min())
print("Maximum:", prices.max())
print("Average:", prices.mean())
print("Total:", prices.sum())
print("Total Bill:", total_bill.sum())
```

```python
unique_prices = np.unique(prices)

print(unique_prices)
print("Number of unique prices:", len(unique_prices))
```
